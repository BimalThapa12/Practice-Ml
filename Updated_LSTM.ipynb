```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer_text=tokenizer.fit_on_texts([faqs])
```

```python
len(tokenizer.word_index)
```

```python
input_sequences = []
for sentence in faqs.split('\n'):
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
input_sequences
```

```python
max_len = max([len(x) for x in input_sequences])

max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding='pre')
```

```python
X = padded_input_sequences[:,:-1]
```

```python
y = padded_input_sequences[:,-1]
```

```python
from tensorflow.keras.utils import to_categorical

vocab_size = len(tokenizer.word_index) + 1  # total words
y = to_categorical(y, num_classes=vocab_size)  # shape: (samples, vocab_size)
```

```python
from keras.models import Sequential
from keras.layers import Embedding, LSTM, Dense

model = Sequential()
model.add(Embedding(input_dim=vocab_size, output_dim=100, input_length=max_len-1))
model.add(LSTM(150,return_sequences=True))
model.add(LSTM(150,return_sequences=True))
model.add(LSTM(150,return_sequences=True))
model.add(LSTM(150))
model.add(Dense(vocab_size, activation='softmax'))  # now matches y


# Compile
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Build model explicitly to see summary
model.build(input_shape=(None, max_len-1))
model.summary()
```

```python
model.fit(X, y, epochs=100)
```

```python
import time
import numpy as np

# Starting text
text = "MR.Bimal"
generated_text = text  # We'll build this

for i in range(50):
    token_text = tokenizer.texts_to_sequences([generated_text])[0]
    padded_token_text = pad_sequences([token_text], maxlen=56, padding='pre')
    pos = np.argmax(model.predict(padded_token_text, verbose=0))

    for word, index in tokenizer.word_index.items():
        if index == pos:
            generated_text += " " + word
            time.sleep(0.5)  # Slower for realism
            break

# Format as a mail
mail = f"""{text},

{generated_text[len(text):].strip().capitalize()}

Best regards,  
Bimal Thapa
"""

print("\n\nGenerated Email:\n")
print(mail)
```
