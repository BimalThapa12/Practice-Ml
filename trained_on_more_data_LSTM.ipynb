```python
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

def sample_with_temperature(preds, temperature=1.0):
    preds = np.asarray(preds).astype('float64')
    preds = np.log(preds + 1e-8) / temperature  # small epsilon to avoid log(0)
    exp_preds = np.exp(preds)
    preds = exp_preds / np.sum(exp_preds)
    probas = np.random.multinomial(1, preds, 1)
    return np.argmax(probas)

def generate_text(model, tokenizer, seed_text, max_seq_len, next_words=50, temperature=0.8):
    output_text = seed_text
    for _ in range(next_words):
        # Convert current text to tokens
        token_list = tokenizer.texts_to_sequences([output_text])[0]
        # Pad tokens to model's input length
        token_list = pad_sequences([token_list], maxlen=max_seq_len-1, padding='pre')
        # Predict next word probabilities
        preds = model.predict(token_list, verbose=0)[0]
        # Sample next word index with temperature
        next_index = sample_with_temperature(preds, temperature)
        # Map index to word
        next_word = tokenizer.index_word.get(next_index, '')
        if next_word == '':
            break  # Stop if no word found (end of vocab)
        # Append next word to output text
        output_text += ' ' + next_word
    return output_text
```
