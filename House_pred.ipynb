```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
class CSVload:
    def __init__(self,filepath):
        self.filepath = filepath
        self.data = None

    def load_data(self):
        try:
            self.data = pd.read_csv(self.filepath)
            print("data Loaded successfully")
        except FileNotFoundError:
            print("file not found please ceck the path and try again")

    def shape_info(self):
        if self.data is not None:
            rows, cols = self.data.shape
            print(f"The dataset has {rows} rows and {cols} columns.")
        else:
            print("Please load the data first using load_data().")

    def show_columns(self):
        if self.data is not None:
            print("🧾 Columns in dataset:")
        for col in self.data.columns:
            print("-", col)
        else:
            print("Please load the data first using load_data().")


    def show_head(self,n=5):
        if self.data is not None:
            print(self.data.head(n))
        else:
            print("data is not loaded")


    def drop(self,x_colm):
        if self.data is not None:
            if isinstance(x_colm, list):
                self.data = self.data.drop(x_colm,axis=1)
                print(f"droped Data {', '.join(x_colm)}")
            else:
                self.data = self.data.drop(columns=[x_colm])
                print("not Dropped")
        else:
            print("data is not defined")

    def info(self):
        if self.data is not None:
            print(self.data.info())
        else:
            print("data is not loaded")

    def find_null(self):
        if self.data is not None:
            print(self.data.isna().sum())
        else:
            print("data is not defined")


    def description(self):
        if self.data is not None:
            print(self.data.describe())
        else:
            print("data is not defined")

    def seperareXandY(self,feature,target):
        X = self.data[feature].values
        y = self.data[target].values

        return X,y
    

    def visualisation(self,x,y):
        import matplotlib.pyplot as plt
        sns.histplot(data=self.data , x=x,y=y)
        plt.show
```
